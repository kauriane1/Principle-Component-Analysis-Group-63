## Step 1: Standardize the Data

```python
# Step 1: Standardize the data
# Formula: z = (x - mean) / std  for each feature column

col_means = np.mean(data_imputed, axis=0)   # mean of each feature
col_stds  = np.std(data_imputed, axis=0, ddof=0)  # std of each feature

# Avoid division by zero for constant columns
col_stds[col_stds == 0] = 1

standardized_data = (data_imputed - col_means) / col_stds

print("Standardized data shape:", standardized_data.shape)
print("\nPost-standardization column means (should be ~0):")
print(np.round(np.mean(standardized_data, axis=0), 4))
print("\nPost-standardization column stds (should be ~1):")
print(np.round(np.std(standardized_data, axis=0), 4))
print("\nFirst 5 rows of standardized data:")
standardized_data[:5]
```

## Why Compute a Covariance Matrix?
The covariance matrix captures pairwise linear relationships between all features: a high
positive value means two indicators move together (e.g. HDI and literacy rate both rise
with development), while a value near zero implies independence. It is also the
mathematical foundation for eigendecomposition in PCA — by finding the directions
(eigenvectors) along which the covariance matrix stretches the data most (largest
eigenvalues), we identify axes that retain the greatest variance, allowing compression
into fewer dimensions without losing critical structural information.

```python
# Step 3: Calculate the Covariance Matrix
# cov(X) = (1/(n-1)) * X_T @ X   where X is already standardized

n = standardized_data.shape[0]
cov_matrix = (standardized_data.T @ standardized_data) / (n - 1)

print("Covariance matrix shape:", cov_matrix.shape)
print("\nCovariance Matrix:")
cov_matrix
```

## Step 4: Eigendecomposition
We decompose the covariance matrix into eigenvalues and eigenvectors using numpy.linalg.eig.
Eigenvalues represent the amount of variance captured by each direction.
Eigenvectors define those directions (the principal components).

```python
import numpy as np

# Data Preparation Steps (consolidated to ensure dependencies are met)
# Assuming 'headers', 'rows' are available from earlier loading

# Identify numeric columns
numeric_cols = headers[2:]

# Extract numeric data from raw rows
numeric_data_raw = np.array([[row[i] for i in range(2, len(headers))] for row in rows], dtype=float)

# Extract Region column and apply label encoding
regions = [row[1] for row in rows]
unique_regions = sorted(set(regions))
region_map     = {r: i for i, r in enumerate(unique_regions)}
region_encoded = np.array([region_map[r] for r in regions], dtype=float)

# Build full numeric matrix: [region_encoded | numeric_cols]
full_numeric = np.column_stack([region_encoded, numeric_data_raw])

# Replace NaN with column mean (imputation)
data_imputed = full_numeric.copy()
for j in range(data_imputed.shape[1]):
    col   = data_imputed[:, j]
    mask  = np.isnan(col)
    if mask.any():
        col_mean         = np.nanmean(col)
        data_imputed[mask, j] = col_mean


# Step 1: Standardize the data (moved here to ensure 'standardized_data' is defined)
# Formula: z = (x - mean) / std  for each feature column

col_means = np.mean(data_imputed, axis=0)   # mean of each feature
col_stds  = np.std(data_imputed, axis=0, ddof=0)  # std of each feature

# Avoid division by zero for constant columns
col_stds[col_stds == 0] = 1

standardized_data = (data_imputed - col_means) / col_stds


# Step 3: Calculate the Covariance Matrix
# cov(X) = (1/(n-1)) * X_T @ X   where X is already standardized

n = standardized_data.shape[0]
cov_matrix = (standardized_data.T @ standardized_data) / (n - 1)

# Step 4: Perform Eigendecomposition
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# Keep only real parts
eigenvalues  = np.real(eigenvalues)
eigenvectors = np.real(eigenvectors)

print("Eigenvalues  shape:", eigenvalues.shape)
print("Eigenvectors shape:", eigenvectors.shape)
print("\nEigenvalues (unsorted):")
print(np.round(eigenvalues, 4))
```

## Step 5: Sort Principal Components by Explained Variance
Eigenvalues are sorted in descending order so that PC1 always captures the most variance.
Each component's individual and cumulative explained variance is printed for transparency.

```python
import numpy as np

# Step 5: Sort eigenvalues and eigenvectors in descending order
sorted_indices    = np.argsort(eigenvalues)[::-1]          # indices from largest to smallest
eigenvalues_sorted = eigenvalues[sorted_indices]
sorted_eigenvectors = eigenvectors[:, sorted_indices]      # columns are eigenvectors

# Explained variance per component
explained_variance_ratio = eigenvalues_sorted / np.sum(eigenvalues_sorted)
cumulative_variance      = np.cumsum(explained_variance_ratio)

print("Sorted eigenvalues:")
print(np.round(eigenvalues_sorted, 4))
print("\nExplained variance ratio per component:")
for i, (ev, cum) in enumerate(zip(explained_variance_ratio, cumulative_variance)):
    print(f"  PC{i+1}: {ev*100:.2f}%  |  Cumulative: {cum*100:.2f}%")

print("\nSorted eigenvectors (each column = one principal component):")
sorted_eigenvectors
```

## Task 2: Dynamic Principal Component Selection
We dynamically select the minimum number of components needed to retain at least 85% of
the total variance. This threshold balances dimensionality reduction against information loss.
The scree plot and cumulative variance curve justify the selection visually.

```python
# Dynamically select the number of components that explain >= 85% of variance
VARIANCE_THRESHOLD = 0.85

num_components = int(np.argmax(cumulative_variance >= VARIANCE_THRESHOLD) + 1)

print(f"Variance threshold : {VARIANCE_THRESHOLD*100:.0f}%")
print(f"Components selected: {num_components}")
print(f"Variance retained  : {cumulative_variance[num_components-1]*100:.2f}%")

# Scree plot
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].bar(range(1, len(explained_variance_ratio)+1),
            explained_variance_ratio * 100, color='steelblue', edgecolor='black')
axes[0].axvline(x=num_components + 0.5, color='red', linestyle='--', label=f'Cut-off at PC{num_components}')
axes[0].set_title('Explained Variance per Component')
axes[0].set_xlabel('Principal Component')
axes[0].set_ylabel('Explained Variance (%)')
axes[0].legend()

axes[1].plot(range(1, len(cumulative_variance)+1),
             cumulative_variance * 100, marker='o', color='darkorange')
axes[1].axhline(y=85, color='red', linestyle='--', label='85% threshold')
axes[1].axvline(x=num_components, color='green', linestyle='--', label=f'PC{num_components}')
axes[1].set_title('Cumulative Explained Variance')
axes[1].set_xlabel('Number of Principal Components')
axes[1].set_ylabel('Cumulative Variance (%)')
axes[1].legend()

plt.tight_layout()
plt.savefig('scree_plot.png', dpi=120)
plt.show()
print("Scree plot saved.")
```

## Step 6: Project Data onto Principal Components
The standardized data is projected onto the top-k eigenvectors to produce the reduced dataset.
This is a linear transformation: reduced = X_standardized · W, where W holds the selected eigenvectors.

```python
# Step 6: Project data onto the selected principal components
# reduced_data = standardized_data @ W  where W = top-k eigenvectors

W            = sorted_eigenvectors[:, :num_components]   # shape: (p, k)
reduced_data = standardized_data @ W                     # shape: (n, k)

print(f"Projection matrix W shape  : {W.shape}")
print(f"Reduced data shape         : {reduced_data.shape}")
print("\nFirst 5 rows of reduced data:")
reduced_data[:5]
```

## Task 3: Performance Optimization
The full PCA pipeline is wrapped in a reusable function and benchmarked against synthetic datasets
ranging from 100 to 10,000 rows. Runtime is measured using Python's time module (no sklearn).
This demonstrates that the implementation scales efficiently to large datasets.

```python
def pca_from_scratch(data, variance_threshold=0.85):
    # 1. Standardize the data
    col_means = np.mean(data, axis=0)
    col_stds = np.std(data, axis=0, ddof=0)
    col_stds[col_stds == 0] = 1
    standardized_data = (data - col_means) / col_stds

    # 2. Calculate the Covariance Matrix
    n = standardized_data.shape[0]
    cov_matrix = (standardized_data.T @ standardized_data) / (n - 1)

    # 3. Perform Eigendecomposition
    eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
    eigenvalues = np.real(eigenvalues)
    eigenvectors = np.real(eigenvectors)

    # 4. Sort eigenvalues and eigenvectors
    sorted_indices = np.argsort(eigenvalues)[::-1]
    eigenvalues_sorted = eigenvalues[sorted_indices]
    sorted_eigenvectors = eigenvectors[:, sorted_indices]

    # Explained variance per component
    explained_variance_ratio = eigenvalues_sorted / np.sum(eigenvalues_sorted)
    cumulative_variance = np.cumsum(explained_variance_ratio)

    # 5. Select number of components
    num_components = int(np.argmax(cumulative_variance >= variance_threshold) + 1)

    # 6. Project data onto the selected principal components
    W = sorted_eigenvectors[:, :num_components]
    reduced_data = standardized_data @ W

    return reduced_data, explained_variance_ratio, num_components


# Run the reusable pca_from_scratch function on the actual dataset
reduced_final, ratios_final, k_final = pca_from_scratch(data_imputed, variance_threshold=0.85)

print(f"Components selected : {k_final}")
print(f"Variance retained   : {np.sum(ratios_final[:k_final])*100:.2f}%")
print(f"Final reduced shape : {reduced_final.shape}")
print("\nFirst 5 rows:")
reduced_final[:5]
```

## Task 3: Performance Benchmarking
The pca_from_scratch() function is tested on synthetic datasets from 100 to 10,000 rows.
Runtime is measured using Python's time module to demonstrate scalability on large datasets.

```python
# Task 3: Benchmark PCA on a large synthetic dataset
import time
import numpy as np
import matplotlib.pyplot as plt

def pca_from_scratch(X, n_components=None, variance_threshold=0.85):
    """
    Full PCA pipeline using numpy only.
    Parameters
    ----------
    X                 : raw numeric numpy array (may contain NaNs)
    n_components      : fixed number of PCs; if None, selected by variance_threshold
    variance_threshold: cumulative variance to retain (used when n_components is None)
    Returns
    -------
    reduced, explained_ratios, n_components_used
    """
    # 1. Impute
    X_imp = X.copy()
    for j in range(X_imp.shape[1]):
        mask = np.isnan(X_imp[:, j])
        if mask.any():
            X_imp[mask, j] = np.nanmean(X_imp[:, j])

    # 2. Standardize
    mu  = np.mean(X_imp, axis=0)
    sig = np.std(X_imp,  axis=0, ddof=0)
    sig[sig == 0] = 1
    X_std = (X_imp - mu) / sig

    # 3. Covariance
    n   = X_std.shape[0]
    cov = (X_std.T @ X_std) / (n - 1)

    # 4. Eigendecomposition
    evals, evecs = np.linalg.eig(cov)
    evals = np.real(evals)
    evecs = np.real(evecs)

    # 5. Sort descending
    idx    = np.argsort(evals)[::-1]
    evals  = evals[idx]
    evecs  = evecs[:, idx]
    ratios = evals / np.sum(evals)
    cumvar = np.cumsum(ratios)

    # 6. Select components
    if n_components is None:
        n_components = int(np.argmax(cumvar >= variance_threshold) + 1)

    # 7. Project
    reduced = X_std @ evecs[:, :n_components]
    return reduced, ratios, n_components


# ── Benchmark on increasingly large synthetic datasets ────────────────────
sizes = [100, 500, 1000, 5000, 10000]
n_features = 10
times = []

print(f"{'Rows':>8}  {'Cols':>6}  {'Time (s)':>10}  {'PCs kept':>8}")
print("-" * 40)

for n_rows in sizes:
    np.random.seed(42)
    X_large = np.random.randn(n_rows, n_features)
    # Inject ~10% NaN
    mask = np.random.rand(n_rows, n_features) < 0.10
    X_large[mask] = np.nan

    t0 = time.perf_counter()
    _, _, k = pca_from_scratch(X_large, variance_threshold=0.85)
    elapsed = time.perf_counter() - t0
    times.append(elapsed)
    print(f"{n_rows:>8}  {n_features:>6}  {elapsed:>10.5f}  {k:>8}")

# Plot benchmark
plt.figure(figsize=(7, 4))
plt.plot(sizes, times, marker='s', color='purple')
plt.title('PCA Runtime vs Dataset Size')
plt.xlabel('Number of Rows')
plt.ylabel('Time (seconds)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('benchmark.png', dpi=120)
plt.show()
print("Benchmark complete.")
```

We chose 85% because it keeps most information while still reducing dimensions. Choosing lower (70%) gives fewer components but loses more detail. Choosing higher (95%) keeps more detail but reduces less.
